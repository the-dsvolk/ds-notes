# using Central Limit theorem for non-stationary analysis


```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
# Data from the last 10 weeks
data = [
    1850000,
    2250000,
    1800000,
    1700000,
    2300000,
    1400000,
    1650000,
    1400000,
    950000,
    1250000,
]
```

```python
# Help me build bar char using matplotlib
plt.bar(range(len(data)), data)
plt.show()
```

CI = x̄ ± 1.96 × (s / sqrt(n))

```python
n = len(data)
mean = np.mean(data)
std_err = np.std(data, ddof=1) / np.sqrt(n)

# --- 1. Z-Score Calculation (95% Confidence) ---
z_val = 1.96
z_moe = z_val * std_err

# --- 2. T-Score Calculation (95% Confidence) ---
t_val = stats.t.ppf(0.975, df=n - 1)
t_moe = t_val * std_err

print(f"Average (X): {mean:,.0f}")
print(f"Z-Score Range (+/-): {z_moe:,.0f}")
print(f"T-Score Range (+/-): {t_moe:,.0f}")
```

Why 0.975 for 95% confidence?
A 95% confidence interval leaves 5% in the tails (2.5% on each side):
     2.5%          95%          2.5%  |-------|==================|-------|      ↑                          ↑   -t_val                     +t_val
So ppf(0.975) gives the t-value where 97.5% of the distribution is to the left.
